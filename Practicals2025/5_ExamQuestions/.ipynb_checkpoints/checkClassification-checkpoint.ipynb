```python
import os

import numpy as np
import matplotlib.pyplot as plt
import scipy.signal

from pyriemann.tangentspace import TangentSpace
from pyriemann.estimation import Covariances
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.manifold import TSNE
from sklearn.model_selection import cross_val_score, StratifiedKFold, cross_val_predict
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from ldakmda import LDA_KM_DA
from sklearn.preprocessing import normalize, StandardScaler
from tdf import TDF
import sys
import mne
import re
from autorank import autorank, plot_stats, create_report, latex_table
import pandas as pd
from MDD import fit_MDD
import seaborn as sn
import warnings
warnings.filterwarnings("ignore")
```

```python
from scipy.signal import spectrogram
from scipy.stats import zscore
from scipy.signal import hilbert
```

```python
subjectNames = []

def loadData(wl=200, windowPerSeconds=20):
    X = []
    EMG = []
    Y = []
    
    subjectNames = []
    
    print("loading data ...")
    for path in os.listdir():
        pattern = r'\b\d+_DataRecording_.*?\b'
        if re.match(pattern, path):
            
            #if not path[:2] in ("01", "02", "03", "04"): continue
            
            print(path)
            subjectNames.append(path)
            emg = np.load(f"{path}/EMG.npy")
            y = np.load(f"{path}/Y.npy")
            timestamps = np.load(f"{path}/Y_timestamps.npy")

            """plt.rcParams['figure.figsize'] = [3, 3]
            t = timestamps[1:] - timestamps[:-1]
            plt.plot(t, linewidth=0.1)
            plt.show()"""

            

            print("classes :", set(list(y)))
            
            sfreq = 500
    
            emg = mne.filter.filter_data(emg, sfreq, 15, 150, verbose="CRITICAL")
            emg = mne.filter.notch_filter(emg, sfreq, 50, notch_widths=5, verbose="CRITICAL")
            emg = mne.filter.notch_filter(emg, sfreq, 100, notch_widths=5, verbose="CRITICAL")
            emg = mne.filter.notch_filter(emg, sfreq, 150, notch_widths=5, verbose="CRITICAL")

            
            for c in range(8):
                emg[c] -= np.mean(emg[c])
                #emg[c] /= np.std(emg[c])
                emg[c] /= np.mean(np.abs(emg[c]))
                #emg[c] /= np.quantile(np.abs(emg[c]), 0.9)


            
            plt.rcParams['figure.figsize'] = [5, 3]
            frequencies, times, Sxx = spectrogram(emg[0], fs=sfreq, nperseg=512)
            mean_spectrum = np.mean(Sxx, axis=1)  # Average across time axis
            plt.plot(frequencies, 10 * np.log10(mean_spectrum))  # Convert to dB
            plt.xlabel('Frequency [Hz]')
            plt.ylabel('Power/Frequency (dB/Hz)')
            plt.title('Average Spectrogram (Power Spectral Density)')
            plt.grid()
            plt.show()

            
            """
            firstSample = 0
            x = []
            for t in timestamps:
                if t > wl:
                    x.append(emg[:, t - wl:t])
                else:
                    firstSample += 1
    
            x = np.array(x)
            y = y[firstSample:]

            x = x[y != -1]
            y = y[y != -1]"""


            labels = np.zeros(len(emg[0]))-2
            last = 0
            for i in range(len(timestamps)):
                labels[last:timestamps[i]] = y[i]
                last = timestamps[i]

            ws = 500 // windowPerSeconds
            x = []
            y = []
            for t in range(wl, len(emg[0]), ws):
                if labels[t] != -1:
                    y.append(labels[t])
                    x.append(emg[:, t-wl:t])
                if labels[t] == -2:
                    break
                
            x = np.array(x)
            print(x.shape)


            

            cov = Covariances(estimator='oas').transform(x)
            ts = TangentSpace('riemann').fit_transform(cov)
            tdf = np.concatenate([f(x) for f in TDF], axis=1)
            tdf = normalize(tdf)
            tdf = StandardScaler().fit_transform(tdf)
            features = np.concatenate([ts, tdf], axis=1)

            
            plt.rcParams['figure.figsize'] = [100, 3]
    
            # cleaning labels
            y = scipy.signal.medfilt(y, kernel_size=3)
            y = scipy.signal.medfilt(y, kernel_size=9)
            y = scipy.signal.medfilt(y, kernel_size=3)

            plt.plot(y)
            
            y2 = LDA_KM_DA(y, features, 0.001).predict(features)
            y2 = scipy.signal.medfilt(y2, kernel_size=3)
            y2 = scipy.signal.medfilt(y2, kernel_size=9)
            y2 = scipy.signal.medfilt(y2, kernel_size=3)
            #y = y2
            
            plt.plot(y2)

            i = 1
            start = 0
            while i < len(y):
                if y2[i] != y2[i-1]:
                    mode = scipy.stats.mode(y[(start+i)//2:i])[0]
                    y2[start:i] = mode
                    start = i
                i+=1
            y = y2

            plt.plot(y)
            plt.show()

            
            X.append(features) 
            EMG.append(x) 
            Y.append(y) 

    
    X = np.array(X, dtype="object")
    Y = np.array(Y, dtype="object")
    EMG = np.array(EMG, dtype="object")
    return X, Y, EMG, subjectNames

X, Y, EMG, subjectNames = loadData(wl=200, windowPerSeconds=20)

print("done")
plt.show()
```

```python
wl = 200

emg = np.load(f"{path}/EMG.npy")
y = yraw = np.load(f"{path}/Y.npy")
timestamps = np.load(f"{path}/Y_timestamps.npy")


emg = mne.filter.filter_data(emg, 500, 5, 150, verbose="CRITICAL")

for c in range(8):
    emg[c] -= np.mean(emg[c])
    emg[c] /= np.std(emg[c])

firstSample = 0
x = []
for t in timestamps:
    if t > wl:
        x.append(emg[:, t - wl:t])
    else:
        firstSample += 1

x = np.array(x)
y = y[firstSample:]

# cleaning labels
y = scipy.signal.medfilt(y, kernel_size=9)

cov = Covariances(estimator='oas').transform(x)
ts = TangentSpace('riemann').fit_transform(cov)
tdf = np.concatenate([f(x) for f in TDF], axis=1)
tdf = normalize(tdf)
tdf = StandardScaler().fit_transform(tdf)
features = np.concatenate([ts, tdf], axis=1)

yold = y
y = LDA_KM_DA(y, features, 0.001).predict(features)
y = scipy.signal.medfilt(y, kernel_size=9)
```
